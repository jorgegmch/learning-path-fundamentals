**EJERCICIO 1: DETERMINAR AHORRO**

Se requiere un algoritmo para determinar cuánto ahorrará una persona en un año, si al final de cada mes deposita variables cantidades de dinero; además, se requiere saber cuánto lleva ahorrado cada mes.

```python
ahorro=0
for mes in range(1, 13):
    deposito=float(input(f"Ingrese el ahorro del mes {mes}: "))
    ahorro=ahorro+deposito
    print(f"Ahorro acumulado hasta el mes {mes}: {ahorro:,.1f}")

print(f"El ahorro total en el año es: {ahorro:,.1f}")
```

**EJERCICIO 3: FACTORIAL DE UN NÚMERO**

Escribe un programa que pida al usuario un número entero positivo y calcule su factorial.

```python
n=int(input("Ingrese un número entero positivo: "))
if n>0:
    fact=1
    for i in range(2, n+1):
        fact=fact*i
    print(f"{n}!={fact}")
elif n==0:
    print("0!=1")
else:
    print("No se puede calcular el factorial porque el número es inválido.")
```

**EJERCICIO 4: GENERAR SERIE FIBONACCI**

Escribe un programa que genere los primeros 10 números de la serie de Fibonacci.

```python
a=0
b=1
print(f"{a}", end=", ")
print(f"{b}", end=", ")
for i in range(2, 10):
    c=a+b
    print(f"{c}", end=", ")
    a=b
    b=c
```

**EJERCICIO 5: GENERAR TRIANGULO DE PASCAL**

Escribe un programa que genere el Triángulo de Pascal hasta la fila n, donde n es un número ingresado por el usuario. El Triángulo de Pascal es un arreglo triangular donde el número en cada posición es la suma de los dos números directamente encima de él en la fila anterior. Nota: No usar vectores.

```python
filas=int(input("Ingrese el número de filas: "))
for i in range(filas):
    numero=1
    for j in range(i+1):
        if j==0 or j==i:
            numero=1
        else:
            numero=numero*(i-j+1)//j
        print(numero, end=" ")
    print()
```

**EJERCICIO 6: MAYORES, IGUALES Y MENORES**

Se requiere un algoritmo para determinar, de N cantidades, cuántas son menores o iguales a cero y cuántas mayores a cero.

```python
n=int(input("Ingrese la cantidad de números: "))
menoresIguales=0
mayores=0
for i in range(n):
    numeroIngresado=float(input(f"Ingrese el número {i+1}: "))
    if numeroIngresado<=0:
        menoresIguales=menoresIguales + 1
    else:
        mayores=mayores+1
print(f"Cantidad de números menores o iguales a 0: {menoresIguales}")
print(f"Cantidad de números mayores a 0: {mayores}")
```
