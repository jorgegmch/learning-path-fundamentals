```python
# Ejemplo 3: Calcular la formula combinatoria de 2 números.
# C(n, k)=n! / (k!*(n-k)!)

def factorial(num1):
    if num1<0:
        return -1
    elif num1==0:
        return 1
    else:
        f=1
        for i in range(1, num1+1):
            f=f*i
        return f

def combi(n, k):
    if n>=k:
        resultado=factorial(n)//(factorial(k)*factorial(n-k))
    else:
        resultado=0
    return resultado

print(f"{combi(5, 3)}")
```

```python
# Ejercicio: Desarrollar el triangulo de pascal.

def factorial(num1):
    if num1<0:
        return -1
    elif num1==0:
        return 1
    else:
        f=1
        for i in range(1, num1+1):
            f=f*i
        return f

def combi(n, k):
    if n>=k:
        resultado=factorial(n)//(factorial(k)*factorial(n-k))
    else:
        resultado=0
    return resultado

def trianguloPascal(n):
    if n<1:
        print("No se puede calcular.")
    else:
        for fila in range(n):
            for col in range(fila+1):
                print(f"{combi(fila, col)}", end=" ")
            print("")
            
trianguloPascal(5)
```

```python
# Ejercicio: Haga 2 funciones que conviertan de Celsius a Fahrenheit 
# y otra que convierta de Fahrenheit a Celsius

def celsius_a_fahrenheit (celsius):
    return (celsius * 9/5) + 32

def fahrenheit_a_celsius (fahrenheit):
    return (fahrenheit - 32) * 5/9

print(f"{celsius_a_fahrenheit(150):.2f}")
print(f"{fahrenheit_a_celsius(150):.2f}")
```
