**ESTRUCTURAS CONDICIONALES SIMPLES**

La estructura se cumple solo cuando la condición es verdadera.

```python
# Ejemplo 1: Mostrar un mensaje solo si una persona es mayor de edad

edad=25
if edad>=18:
    print("Usted es MAYOR de edad.")
    
# La condición es verdadera porque la variable de edad es 25 y el mensaje solo se imprimirá si la edad de la persona es mayor o igual a 18.
```

```python
# Ejemplo 2.1: Mostrar un mensaje solo si una persona es menor de edad.

edad=25
if edad<=17:
    print("Utes es MENOR de edad.")
    
# La condición es falsa porque la variable de edad es 25 y solo imprimirá el mensaje si la edad de la persona es menor o igual a 17.
```

```python
# Ejemplo 2.2: Mostrar un mensaje solo si una persona es menor de edad.

edad=25
if not edad>=18:
    print("Utes es MENOR de edad.")
    
# La condición sigue siendo falsa porque la variable de edad es 25 y solo imprimirá el mensaje si la edad de la persona es menor o igual a 17.
```

```python
# Ejemplo 1: Simule el lanzamiento de 2 dados. El usuario debe intentar adivinar que número cayó.
# Si adivina, debe mostrar el mensaje: "¡Felicidades! Has adivinado."
# Si no adivina y su número es mayor que los dados, el mensaje debe ser: "El número fue menor."
# Si no adivina y su número es menor que los dados, el mensaje debe ser: "El número fue mayor."

import random

dado1=random.randint(1, 6)
dado2=random.randint(1, 6)
dados=dado1+dado2

adivina=int(input("Ingresa el número que crees que saldra en los dados (1-12): "))
if adivina<1 or adivina>12:
    print("Número inválido.")
elif adivina==dados:
    print("¡Felicidades! Has adivinado.")
elif adivina>dados:
    print("El número fue menor.")
else:
    print("El número fue mayor.")
```

**EJERCICIO 2**

Dado el nombre del estudiante y la calificación cuantitativa de una evaluación (0-100), se pide hallar la calificación cualitativa, de acuerdo a la siguiente información.

*Rango Evaluación*

0 - 59

60 - 79

80 - 89

90 - 100

*Evaluación cualitativa*

D

C

B

A

Se pide visualizar, nombre, calificación cuantitativa y cualitativa.

```python
nombreEstudiante=input("Ingrese el nombre del estudiante: ")
calificacion=int(input("Ingrese la nota o califcación cuantitativa del estudiante (0-100): "))

if calificacion>0 and calificacion<=59:
    print(f"El estudiante {nombreEstudiante} obtuvo una calificación cuantitativa de {calificacion}, lo que es igual a una D")
elif calificacion>59 and calificacion<=79:
    print(f"El estudiante {nombreEstudiante} obtuvo una calificación cuantitativa de {calificacion}, lo que es igual a una C")
elif calificacion>79 and calificacion<=89:
    print(f"El estudiante {nombreEstudiante} obtuvo una calificación cuantitativa de {calificacion}, lo que es igual a una B")
elif calificacion>89 and calificacion<=100:
    print(f"El estudiante {nombreEstudiante} obtuvo una calificación cuantitativa de {calificacion}, lo que es igual a una A")
else:
    print("Calificación incorrecta.")
```

**ESTRUCTURA MULTIPLE MATCH-CASE**

Es una forma elegante de comparar un valor contra multiples case literales. Python 3.10+

match valor:

    case patron_1:

        # Bloque A

    case patron_2:

        # Bloque B

    case patron_3:

        # Bloque C

    ...

    case patron_10:

        # Bloque J

    case _:
    
        # Bloque o caso comodín (opcional)

```python
# Ejemplo 1: Suponga que un sistema antiguo nos da el día de la semana con un número.
# 1. Lunes / 2. Martes / 3. Miercoles / 4. Jueves / 5. Viernes / 6. Sabado / 7. Domingo
# Construya un programa que traduzca ese número al día de la semana correspondiente.

import random

diaSem=random.randint(1, 7)

print(f"Día de la semana: {diaSem}")
match diaSem:
    case 1:
        print("Lunes.")
    case 2:
        print("Martes.")
    case 3:
        print("Miercoles.")
    case 4:
        print("Jueves.")
    case 5:
        print("Viernes.")
    case 6:
        print("Sabado.")
    case 7:
        print("Domingo.")
    case _:
        print("Valor no válido.")
```

**EJERCICIO 3**

Índice de Masa Corporal (IMC): Pide al usuario su peso en kg y su altura en metros. Calcula el IMC (peso/altura ^2). Clasifica el resultado según la siguiente tabla e imprime la categoría:

* Menos de 18.5: Bajo peso

* 18.5 - 24.9: Peso normal

* 25.0 - 29.9: Sobrepeso

* 30.0 o más: Obesidad

```python
peso=float(input("Ingrese su peso en kilogramos: "))
altura=float(input("Ingrese su altura en metros: "))
imc=peso/(altura**2)

if imc<18.5:
    print(f"Usted se encuentra por debajo de su peso ideal y su IMC es: {imc:.2f}")
elif imc>=18.5 and imc<24.9:
    print(f"Su peso es normal y su IMC es: {imc:.2f}")
elif imc>=24.9 and imc<29.9:
    print(f"Usted tiene sobrepeso y su IMC es: {imc:.2f}")
else:
    print(f"Usted tiene obesidad y su IMC es: {imc:.2f}")
```
