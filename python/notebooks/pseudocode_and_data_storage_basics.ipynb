**EJERCICIO #1 - CALCULAR NÓMINA**

La empresa ACME desea calcular el valor de la nómina de un empleado, tanto el  sueldo bruto como el sueldo neto. El sueldo bruto se calcula a partir del valor de la hora y la cantidad de horas trabajadas, a esto se le descuenta el valor de la EPS que es el 4% y el valor de la Pensión que es el 4%. El sueldo neto es el sueldo bruto menos los descuentos. Mostrar al final, el valor del sueldo bruto, cada uno de los descuentos y el  valor del sueldo neto. Para este ejercicio el valor de la hora es de $20.000.

*Entrada:* horasTrabajadas

*Proceso:* sueldoBruto=horasTrabajadas x 20000 | descuentoSalud=sueldoBruto x 0.04 | descuentoPension=sueldoBruto x 0.04 | 
sueldoNeto=sueldoBruto-(descuentoSalud+descuentoPension)

*Salida:* sueldoBruto | Descuento por salud y pensión | sueldoNeto

**ALGORITMO EN PSEUDOCODIGO**

*Inicio*

1. Leer horasTrabajadas.
2. Definir variable de valorHora.
3. Calcular el valor del sueldoBruto.
4. Calcular el valor de descuentoSalud.
5. Calcular el valor de descuentoPension.
6. Calcular el valor del sueldoNeto.
5. Imprimir el sueldoBruto.
6. Imprimir descuentos por salud.
7. Imprimir descuento por pensión.
8. Imprimir sueldoNeto.

*Fin*

```python
horasTrabajadas=float(input("Ingrese el total de horas trabajadas: "))
valorHora=20000
sueldoBruto=horasTrabajadas*valorHora
descuentoSalud=sueldoBruto*0.04
descuentoPension=sueldoBruto*0.04
sueldoNeto=sueldoBruto-(descuentoSalud+descuentoPension)
print(f"Sueldo bruto= {sueldoBruto:,.1f}")
print(f"Descuento por salud(EPS)= -{descuentoSalud:,.1f}")
print(f"Descuento por pensión= -{descuentoPension:,.1f}")
print(f"Sueldo neto= {sueldoNeto:,.1f}")
```

**EJERCICIO #2 - CALCULAR CURVA DE 8 DE UNA NOTA**

*Entrada:* notaAcademica

*Salida:* notaFinal

**ALGORITMO EN PSEUDOCODIGO**

*Inicio*

2. Verificar que esté en el rango válido (0.0 a 5.0).

Sí

2. Calcular curva de 8 para notaAcedemica.

**EJERCICIO #3 - CALCULAR NUEVA HORA**

Hacer un algoritmo que, dada una hora con minutos y minutos adicionales, calcule la nueva hora. Asuma que no se pasa de las 23:59.

*Ejemplo:* Hora: 10:45 | Minutos: 20 | Nueva Hora: 11:05

*Entrada:* horaInicio | minActuales | minAdicionales

*Proceso:* minTotales=minActuales+minAdicionales | horaNueva=horaInicio+(minTotales//60) | minNuevos=minTotales%60

*Salida:* horaNueva:minNuevos

**ALGORITMO EN PSEUDOCODIGO**

*Inicio*

1. Leer horaInicio.
2. Leer minutosActuales.
3. Leer minAdicionales
4. Calcular minTotales.

Sí minTotales es mayor o igual que 60.

6. Calcular horaNueva.
7. Calcular minNuevos

Sino

9. horaNueva es igual a horaInicio.
10. minNuevos es igual a minTotales.
11. Imprimir horaNueva y minNuevos.

*Fin*

```python
horaInicio=int(input("Ingrese la hora actual entre 0-23: "))
minActuales=int(input("Ingrese los minutos actuales entre 0-59: "))
minAdicionales=int(input("Ingrese los minutos que desea adicionar: "))
minTotales=minActuales+minAdicionales
if minTotales>=60:
    horaNueva=horaInicio+(minTotales//60)
    minNuevos=minTotales%60
else:
    horaNueva=horaInicio
    minNuevos=minTotales
print(f"La hora que desea conocer corresponde a las: {horaNueva}:{minNuevos}")
```

**EJERCICIO #4 - COSTO PEDIDO DE ARTÍCULO**

Elabore un algoritmo en pseudocódigo que imprima el costo de un pedido de un artículo del cual se tiene la descripción, la cantidad pedida y el precio unitario. Debe  mostrar la descripción del producto, el valor antes del IVA, el IVA y el valor con IVA. Nota: El IVA es el 20% del valor bruto.

**ALGORITMO EN PSEUDOCODIGO**

*Inicio*

1. Definir variables de descripcionProducto, cantidadSolicitada, valorUnidad y porcentajeIVA.
2. Calcular valorTotalSinIVA, valorUnidadConIVA y totalPagar.
3. Imprimir "COSTO DEL PEDIDO".
4. Imprimir "descripcionProducto".
5. Imprimir "cantidadSolicitada"
6. Imprimir "valorUnidad"
7. Imprimir "valorTotalSinIVA"
8. Imprimir "IVA"
9. Imprimir "valorUnidadConIVA"
10. Imprimir "totalPagar"

*Fin*

```python
descripcionProducto="Silla mecedora"
cantidadSolicitada=3
valorUnidad=150000
visualPorcentajeIVA=20
conversionProductoIVA=0.20
valorTotalSinIVA=cantidadSolicitada*valorUnidad
valorUnidadConIVA=(valorUnidad*conversionProductoIVA)+valorUnidad
totalPagar=cantidadSolicitada*valorUnidadConIVA
print("COSTOS DEL PEDIDO")
print(f"Producto: {descripcionProducto}")
print(f"Cantidad: {cantidadSolicitada}")
print(f"Precio por unidad sin IVA: {valorUnidad:,.1f}")
print(f"Precio total sin IVA: {valorTotalSinIVA:,.1f}")
print(f"IVA {visualPorcentajeIVA}%")
print(f"Precio por unidad con IVA: {valorUnidadConIVA:,.1f}")
print(f"Total a pagar: {totalPagar:,.1f}")
```
