## Algoritmo de calculo del area de un circulo
Diseñe un algoritmo y un progrma ane Python que calcule el area de un circulo

**Entrada:** Radio 
**Proceso:** areaCirculo = pi*radio^2
**Salida:** area circulo

## Algoritmo
**Inicio**
1. Leer radio
2. areaCirculo=3.1416*radio**2
3. Escribir areaCirculo
**Fin**

```python
radio=float(input("Ingrese el valor del radio: "))
areaCirculo=3.1416*radio**2
print(f"El area del circulo: {areaCirculo:.2f}")
```

## Ejercicio 1
Diseñar un algoritmo que lea cuatro números y calcule e imprima su producto, suma y su media aritmetica

**Entrada:** numero1, numero2, numero3, numero4
**Proceso:** promedio=(num1+num2+num3+num4)/4
**Salida:** Suma y media aritmetica(promedio)

## Algoritmo
**Inicio**
1. Leer num1, num2, num3, num4
2. Sumar los 4 números y dividirlos entre 4 para calcular su area aritmetica(promedio)
3. Imprimir el resultuado de la suma y de la media aritmetica(promedio)
**Fin**

```python
num1=float(input("Ingrese el valor del primer número: "))
num2=float(input("Ingrese el valor del segundo número: "))
num3=float(input("Ingrese el valor del tercer número: "))
num4=float(input("Ingrese el valor del cuarto número: "))
multiplicacion=num1*num2*num3*num4
suma=num1+num2+num3+num4
promedio=(num1+num2+num3+num4)/4
print(f"El producto de los 4 números es igual a: {multiplicacion}, su suma es: {suma} y su media aritmetica: {promedio}")
```

## Ejercicio 2
Diseñar un algoritmo y programa que lea el peso de una persona en libras y lo devuelva en kilogramos y gramos.

**Entrada:** pesoLibras
**Proceso:** pesoKilogramos=pesoLibras*0.453592 
             pesoGramos=pesoLibras*453.592
**Salida:** Peso en kilogramos y gramos

## Algoritmo
**Inicio**
1. Leer pesoLibras
2. Convertir pesoLibras a pesoKilogramos y pesoGramos
3. Imprimir el resultado de pesoKilogramos y pesoGramos

## Ejercicio 3
Diseñar un algoritmo y programa que lea la altura de una persona en pies y su peso en libras y calcule su IMC (índice de masa corporal)
IMC=peso(kg)/(altura(m)^2)

**Entrada:** pesoLibras y alturaPies
**Proceso:** pesoKilogramos=pesoLibras*0.453592
             alturaMetros=alturaPies*0.3048
             imc=pesoKilogramos/(alturaMetros**2)
**Salida:** IMC (Indice de masa corporal)

## Algoritmo
**Inicio**
1. Leer pesoLibras y alturaPies
2. Convertir pesoLibras a pesoKilogramos y alturaPies a alturaMetros
3. Con los resultados de la conversión anterior, calcular IMC con su formula
4. Imprimir el IMC
**Fin**

```python
pesoLibras=float(input("Ingrese su peso en Libras: "))
alturaPies=float(input("Ingrese su altura en pies: "))
pesoKilogramos=pesoLibras*0.453592
alturaMetros=alturaPies*0.3048
imc=pesoKilogramos/(alturaMetros**2)
print(f"Su IMC es: {imc:.1f}")
```
