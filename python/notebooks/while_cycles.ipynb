## CICLOS WHILE

Son bucles que se ejecutan mientras la condición sea verdadera.

while condición: = mientras condición haga

```python
# Ejemplo 1: mostrar con un ciclo while los números del 1 al 10.

num=1
while num<=10:
    print(f"{num}", end=", ")
    num=num+1
```

```python
# Ejemplo 2:
# Fibonnaci fue un matematico que modelo el crecimiento de los conejos mediante la sucesión:
# 1, 1, 2, 3, 5, 8, ...
# donde fib(1)=1, fib(2)=1, fib(3)=2, fib(k)=fib(k-2)+fib(k-1)
# Calcule el n-esimo termino de la función fibonacci que el usuario ingrese.

n=int(input("Ingrese el término de Fibonacci que quiere hallar: "))
if n<1:
    print("Error. Entrada inválida.")
elif n==1:
    print("1")
elif n==2:
    print("1, 1")
else:
    ant=1
    sig=1
    temp=0
    print("1, 1", end=", ")
    for i in range(3, n+1):
        temp=sig
        sig=ant+sig
        ant=temp
        print(f"{sig}", end=", ")
```

```python
# Ejemplo 3: Codigo que válide que el usuario haya digitado "F" para femenio y "M" para masculino.
# Mostrar un mensaje de que se equivocó e intente nuevamente.

sexo=input("¿Sexo (F o M)?")
while (sexo!="f" and sexo!="F") and (sexo!="m" and sexo!="M"):
    print("Error. Intente nuevamente.")
    sexo=input("¿Sexo (F o M)?")
    
print("Ha ingresado el sexo correctamente.")
```

```python
# Ejemplo 4: Un profesor ingresa cierta cantidad de notas (notas de 1 a 10) y quiere saber,
# cual es el promedio, cual es la mejor y la peor nota. 
# Cuando no tenga más notas, ingresará un valor menor que 1 para terminar.

suma=0
nota=int(input("Ingrese una nota: "))
cantNotas=1
mejorNota=nota
peorNota=nota
while nota>=1:
    suma=suma+nota
    
    nota=int(input("Ingrese una nota: "))
    if nota>=1:
        cantNotas=cantNotas+1
        if nota>mejorNota:
            mejorNota=nota
        elif nota<peorNota:
            peorNota=nota
            
promedio=suma/cantNotas
print(f"El promedio de nota es: {promedio:.2f}")
print(f"La mejor nota es: {mejorNota}")
print(f"La menor nota es: {peorNota}")
```
