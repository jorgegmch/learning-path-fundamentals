## CONTADORES

Me sirve para saber cuantos elementos hay.

* Empiezan en 0: cont=0

* Se incrementan de 1 en 1: cont+1

```python
# Ejemplo 1: Cree un programa que genere 100 números aleatorios (rango de 1 a 1000)
# y me indique cuantos y que porcentaje de esos son impares.

import random

cont=0
for i in range (100):
    num=random.randint(1, 1000)
    if num%2!=0:
        cont=cont+1
        
porc=cont/100*100
print(f"Cantidad de números impares: {cont}")
print(f"Porcentaje de números impares: {porc:.2f}%")
```

```python
# Ejemplo 2:
# En un salón hay 10 estudiantes, cuente cuantos son mayores de edad,
# cuantos son menores de edad, cuantos son mujeres y cuantos son hombres.
# Si hay más hombres que mujeres mostrar el mensaje: "Se debe aumentar la cuota femenina".

CANTEST=4
contMayEdad=0
contMenEdad=0
contMujeres=0
contHombres=0

for i in range(CANTEST):
    print(f"Datos del estudiante # {i+1}")
    edad=int(input(f"\t¿Edad? # {i+1}"))
    sexo=input(f"\t¿Sexo (F o M)? # {i+1}")
    
    if edad>=18:
        contMayEdad=contMayEdad+1
    else:
        contMenEdad=contMenEdad+1
    
    if sexo=="f" or sexo=="F":
        contMujeres=contMujeres+1
    else:
        contHombres=contHombres+1
        
print(f"Estudiantes mayores de edad: {contMayEdad}")
print(f"Estudiantes menores de edad: {contMenEdad}")
print(f"Estudiantes mujeres: {contMujeres}")
print(f"Estudiantes hombres: {contHombres}")
print("Se debe aumentar la cuota femenina.") if contHombres>contMujeres else print("")
```
