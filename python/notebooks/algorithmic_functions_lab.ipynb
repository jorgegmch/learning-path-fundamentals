**EJERCICIO 1**

Diseñe un algoritmo con funciones y procedimientos que indique el valor del descuento de un artículo el cual es del 5% solo si el artículo tiene un costo superior a $150.000.

```python
def descuento():
    articulo=input("Ingrese el nombre del artículo: ")
    precio=int(input("Ingrese el precio del artículo sin comas ni puntos: "))
    if precio>150000:
        porcentaje=5/100
        descArt=precio*porcentaje
        precioFinal=precio-descArt
        print(f"Artículo: {articulo}")
        print(f"Precio original = {precio:,}")
        print("DESCUENTO ACTIVADO DEL 5%")
        print(f"Precio final = {precioFinal:,}")
    else:
        print(f"El artículo '{articulo}' NO TIENE DESCUENTO, debe pagar el precio original de: {precio:,}")

descuento()
```

**EJERCICIOS 2**

Diseñe un algoritmo con funciones y procedimientos que dado un número entero entre 0 y 20 diga si es o no un número primo. Recuerde que los números primos menores o iguales a 20 son: 2, 3, 5, 7, 11, 13, 17, 19

```python
def primos():
    num=int(input("Ingrese un número entero positivo entre 0-20: "))
    if num==2:
        print(f"El {num} es el único número par entre 0-20 que es PRIMO.")
    elif num>=3 and num<=20:
        if not num%2==0:
            print(f"El número {num} es PRIMO.")
        else:
            print(f"El número {num} NO es PRIMO.")
    elif num<=1:
        print("Los números menores o iguales a 1 NO SON PRIMOS.")
    else:
        print("Debe ingresar un número entero positivo entre 0-20")
primos()
```

**EJERCICIO 3**

Diseñe un pseudocódigo que permita registrar 10 estudiantes (código, nombre). Además del registro, deberá:

    1. Asignar 3 notas parciales a cada estudiante.

    2. Registrar el número de inasistencias de cada estudiante.

    3. Calcular nota final.

    4. Mostrar el listado de estudiantes.

La nota final será calculada por promedio aritmético de las notas parciales y deberá disminuir acorde a las inasistencias, así:

    Igual o mayor a 10 y menor a 15 inasistencias: -0.5

    Igual o mayor a 15 inasistencias: -1

Al elegir la opción 4, deberá mostrar la lista de estudiantes con sus notas parciales, inasistencias y la nota final. Ejemplo:



```python
def inasis(promedio, inasistencia):
    if inasistencia >= 10 and inasistencia < 15:
        promedio = promedio - 0.5
    elif inasistencia >= 15:
        promedio = promedio - 1
    return promedio

for e in range(10):
    nomEstudiante = input(f"Ingrese el nombre del estudiante #{e+1}: ")
    inasistencia = int(input(f"Ingrese la cantidad de inasistencias del estudiante #{e+1}: "))

    suma = 0
    for n in range(3):
        nota = float(input(f"Ingrese la nota #{n+1} entre 1-10: "))
        suma = suma + nota

    promedio = suma / 3
    notaFinal = inasis(promedio, inasistencia)

    print("\n~~ RESULTADOS DEL ESTUDIANTE ~~\n")
    print(f"COD: {e+1}")
    print(f"NOMBRE: {nomEstudiante}")
    print(f"INASISTENCIAS: {inasistencia}")
    print(f"NOTA FINAL: {notaFinal:.2f}")
    input("\nPresione ENTER para continuar...\n")
```
