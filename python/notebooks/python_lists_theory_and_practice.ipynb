# LISTAS EN PYTHON

**¿Qué es una lista?**

Una lista es una estructura de datos que permite almacenar múltiples elementos en una sola variable. Es como tener una caja donde puedes guardar varios objetos y cada uno tiene una posición específica.

**Características principales de las listas**

Ordenadas: Los elementos mantienen el orden en que se agregaron.

Indexadas: Cada elemento tiene una posición (índice) que comienza en 0.

Mutables: Se pueden modificar después de crearlas (agregar, eliminar, cambiar elementos).

Heterogéneas: Pueden contener diferentes tipos de datos (números, cadenas, booleanos).

Dinámicas: Pueden crecer o reducir su tamaño durante la ejecución.


```python
# Ejemplo 1: Sintaxis básica

# Lista vacía
mi_lista = []
# Lista con elementos
numeros = [1, 2, 3, 4, 5]
nombres = ["Ana", "Carlos", "María"]
mixta = [1, "Hola", True, 3.14]
print(numeros) # [1, 2, 3, 4, 5]
print(nombres) # ['Ana', 'Carlos', 'María']
print(mixta) # [1, 'Hola', True, 3.14]
```

```python
# Ejemplo 2: Listas con diferentes tipos de datos

# Lista de números enteros
edades = [18, 25, 30, 22, 35]
# Lista de números decimales
precios = [10.50, 25.99, 5.00, 100.75]
# Lista de cadenas
ciudades = ["Bogotá", "Medellín", "Cali", "Barranquilla"]
# Lista de booleanos
estados = [True, False, True, True, False]
# Lista mixta
informacion = ["Juan", 25, True, 175.5, "Ingeniero"]
```

## Acceso a elementos (Indexación)

**Índices positivos**

Los índices comienzan en 0 para el primer elemento.

```python
# Ejemplo 1:

frutas = ["manzana", "banana", "naranja", "uva", "pera"]
print(frutas[0]) # manzana (primer elemento)
print(frutas[1]) # banana (segundo elemento)
print(frutas[2]) # naranja (tercer elemento)
print(frutas[4]) # pera (quinto elemento)
# print(frutas[5]) # Error! Índice fuera de rango
```

**Índices negativos**

Permiten acceder desde el final de la lista hacia atrás.

```python
colores = ["rojo", "azul", "verde", "amarillo", "negro"]
print(colores[-1]) # negro (último elemento)
print(colores[-2]) # amarillo (penúltimo elemento)
print(colores[-3]) # verde
print(colores[-5]) # rojo (primer elemento)
```

## Propiedades básicas de las listas

**Función len() - Longitud de la lista**

```python
numeros = [10, 20, 30, 40, 50]
print(len(numeros)) # 5
nombres = ["Pedro", "Ana", "Luis"]
print(len(nombres)) # 3
lista_vacia = []
print(len(lista_vacia)) # 0
```

**Verificar si un elemento está en la lista ( in )**

```python
animales = ["perro", "gato", "pájaro", "pez"]
print("perro" in animales) # True
print("elefante" in animales) # False
print("gato" in animales) # True
# También funciona con números
numeros = [1, 3, 5, 7, 9]
print(5 in numeros) # True
print(4 in numeros) # False
```

**Verificar si un elemento NO está en la lista ( not in )**

```python
materias = ["matemáticas", "español", "ciencias", "historia"]
print("inglés" not in materias) # True
print("matemáticas" not in materias) # False
```

### Modificación de listas

**Cambiar elementos existentes**

```python
calificaciones = [7, 8, 6, 9, 5]
print("Original:", calificaciones)
# Cambiar un elemento específico
calificaciones[2] = 10 # Cambiar el tercer elemento
print("Modificada:", calificaciones) # [7, 8, 10, 9, 5]
# Cambiar múltiples elementos
calificaciones[0] = 9
calificaciones[4] = 8
print("Final:", calificaciones) # [9, 8, 10, 9, 8]
```

**.append() = Se agrega un elemento al final de la lista.**

```python
compras = ["pan", "leche", "huevos"]
print("Lista inicial:", compras)
compras.append("queso")
print("Después de agregar queso:", compras)
compras.append("mantequilla")
print("Después de agregar mantequilla:", compras)
# Resultado: ['pan', 'leche', 'huevos', 'queso', 'mantequilla']
```

**.insert(indice, elemento) = Insertar elementos en posición específica.**

```python
colores = ["rojo", "azul", "verde"]
print("Original:", colores)
# insert(posición, elemento)
colores.insert(1, "amarillo") # Insertar en posición 1
print("Después de insertar:", colores) # ['rojo', 'amarillo', 'azul', 'verde']
colores.insert(0, "negro") # Insertar al inicio
print("Con negro al inicio:", colores) # ['negro', 'rojo', 'amarillo', 'azul', 'verde']
```

**.remove(indice) = Eliminar por valor.**

```python
frutas = ["manzana", "banana", "naranja", "banana", "uva"]
print("Original:", frutas)
frutas.remove("banana") # Elimina la PRIMERA aparición
print("Sin primera banana:", frutas) # ['manzana', 'naranja', 'banana', 'uva']
frutas.remove("uva")
print("Sin uva:", frutas) # ['manzana', 'naranja', 'banana']
# frutas.remove("kiwi") # Error! kiwi no está en la lista
```

.pop(indice) = Eliminar por posición.

```python
numeros = [10, 20, 30, 40, 50]
print("Original:", numeros)
# pop() sin argumentos elimina el último elemento
ultimo = numeros.pop()
print("Eliminé:", ultimo) # 50
print("Lista actual:", numeros) # [10, 20, 30, 40]
# pop(índice) elimina elemento en posición específica
segundo = numeros.pop(1)
print("Eliminé:", segundo) # 20
print("Lista final:", numeros) # [10, 30, 40]
```

**.clear() = Vaciar toda la lista y la guarda como una variable temporal por si desea volver a utilizarla en otro momento.**

```python
temporales = [1, 2, 3, 4, 5]
print("Antes:", temporales)
temporales.clear()
print("Después:", temporales) # []
```

**.count(indice) = Cuenta cuantas veces aparece un elemento en la lista.**

```python
numeros = [1, 2, 3, 2, 4, 2, 5]
print(numeros.count(2)) # 3 (aparece 3 veces)
print(numeros.count(1)) # 1 (aparece 1 vez)
print(numeros.count(9)) # 0 (no aparece)
letras = ['a', 'b', 'a', 'c', 'a']
print(letras.count('a')) # 3
```

**.index() = Encuentra la posición o indice de un elemento.**

```python
colores = ["rojo", "azul", "verde", "amarillo", "azul"]
print(colores.index("verde")) # 2
print(colores.index("azul")) # 1 (primera aparición)
# print(colores.index("negro")) # Error! negro no está en la lista
# Búsqueda segura
color_buscar = "morado"
if color_buscar in colores:
    posicion = colores.index(color_buscar)
    print(f"{color_buscar} está en posición {posicion}")
else:
    print(f"{color_buscar} no está en la lista")
```

**.sort() = Ordena de forma permanente los datos en ascendente.**

**.sort(reverse=True) = Ordena de forma permanente los datos en descendente.**

```python
numeros = [5, 2, 8, 1, 9, 3]
print("Original:", numeros)
numeros.sort() # Orden ascendente
print("Ascendente:", numeros) # [1, 2, 3, 5, 8, 9]
numeros.sort(reverse=True) # Orden descendente
print("Descendente:", numeros) # [9, 8, 5, 3, 2, 1]
# Con cadenas (orden alfabético)
nombres = ["Carlos", "Ana", "Beatriz", "David"]
nombres.sort()
print("Nombres ordenados:", nombres) # ['Ana', 'Beatriz', 'Carlos', 'David']
```

**sorted(lista) = Se le asigna una variable al sorted y dentro del parentesis el nombre de la variable de la lista y la ordena sin afectar a la original.**

```python
original = [5, 2, 8, 1, 9]
ordenada = sorted(original)
print("Original:", original) # [5, 2, 8, 1, 9]
print("Ordenada:", ordenada) # [1, 2, 5, 8, 9]
```

**.reverse() = Invierte el orden de la lista.**

```python
numeros = [1, 2, 3, 4, 5]
print("Original:", numeros)
numeros.reverse()
print("Invertida:", numeros) # [5, 4, 3, 2, 1]
```

**.copy() = Copia una lista y se pueden modificar de forma indidivual.**

```python
original = [1, 2, 3, 4, 5]
copia = original.copy()
# Modificar la copia no afecta el original
copia.append(6)
copia[0] = 100
print("Original:", original) # [1, 2, 3, 4, 5]
print("Copia:", copia) # [100, 2, 3, 4, 5, 6]
```

### Recorrer listas con for

Para recorrer una lista con for, se usa *for "iterante" in "lista"*, pero también podemos recorrer lista con indices usando *for "iterante" in range(len("lista"))*.

También podemos recorrer la lista con enumerate() para que me diga cuantos elementos hay en la lista y me los imprima enumerandolos, su sintaxis sería: *for "indice", "elemento" in enumerate("lista", "valor desde donde inicia")*

Para separar una lista usamos *"lista"[inicio:fin:paso]*
