**EJERCICIO 1: CONVERTIDOR DE TEMPERATURA**

Pide al usuario que ingrese una temperatura y una  unidad ('C' para Celsius, 'F' para Fahrenheit). Si la unidad es 'C', conviértela a  Fahrenheit e imprímela. Si es 'F', conviértela a Celsius e imprímela. (F = C * 9/5 +  32; C = (F - 32) * 5/9). 


```python
unidad=input("Ingrese 'C' para convertir desde Celsius o 'F' para hacerlo desde Fahrenheit").lower()

if unidad=="c":
    tempCelsius=float(input("Ingrese el valor de la temperatura en Celsius: "))
    f=(tempCelsius * 9/5) + 32
    print(f"{tempCelsius:.1f} grados Celsius (°C) es igual a {f:.2f} grados Fahrenheit (°F).")
elif unidad=="f":
    tempFahrenheit=float(input("Ingrese el valor de la temperatura en Fahrenheit: "))
    c=(tempFahrenheit - 32) * 5/9
    print(f"{tempFahrenheit:.1f} grados Fahrenheit (°F) es igual a {c:.2f} grados Celsius (°C).")
else:
    print("Opción no válida.")
```

**EJERCICIO 2: CALCULADORA DE AÑO BISIESTO**

Pide al usuario un año. Un año es bisiesto si cumple una de estas dos condiciones: es divisible por 4 pero no por 100, a menos que también sea divisible por 400. El programa debe imprimir si el año es bisiesto o no. 

```python
año=int(input("Ingrese un año sin puntos ni comas: "))

if (año//4)*4==año and not (año//100)*100==año:
    print(f"El {año} es un año bisiesto.")
elif (año//4)*4==año and (año//100)*100==año and (año//400)*400==año:
    print(f"El {año} es un año bisiesto.")
else:
    print(f"El {año} no es un año bisiesto.")
```

**EJERCICIO 3: VALIDADOR DE FECHAS** 

Pide al usuario que ingrese un día, mes y año como números. El programa debe determinar si la fecha es válida. Considera meses de 30/31 días y que febrero tiene 29 días en año bisiesto y 28 en uno normal. El  programa debe imprimir "Fecha válida" o "Fecha inválida".

```python
dia=int(input("Ingrese un día en números (1-31): "))
mes=int(input("Ingrese un mes en números (1-12): "))
año=int(input("Ingrese un año sin puntos ni comas: "))

# año bisiesto o no bisiesto
añoDiv4=(año//4)*4==año
añoDiv100=(año//100)*100==año
añoDiv400=(año//400)*400==año

if dia<1 or dia>31 or mes<1 or mes>12:
    print("Fecha inválida.")
else:
    if mes==1 or mes==3 or mes==5 or mes==7 or mes==8 or mes==10 or mes==12:
        if dia>=1 and dia<=31:
            print("Fecha válida.")
        else:
            print("Fecha inválida.")
    elif mes==4 and mes==6 and mes==9 and mes==11:
        if dia>=1 and dia<=30:
            print("Fecha válida.")
        else:
            print("Fecha inválida.")
    elif mes==2:
        if (añoDiv4 and not añoDiv100) or añoDiv400:
            if dia>=1 and dia<=29:
                print("Fecha válida.")
            else:
                print("Fecha inválida.")
        else:
            if dia>=1 and dia<=28:
                print("Fecha válida.")
            else:
                print("Fecha inválida.")
```
