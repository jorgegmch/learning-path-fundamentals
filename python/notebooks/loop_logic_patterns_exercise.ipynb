**EJERCICIO EN CLASE CON CONDICIONALES**

Pedir al usuario que ingrese 3 números enteros positivos y organizarlos de menor a mayor.

```python
num1=int(input("Ingrese un primer número entero positivo: "))
num2=int(input("Ingrese un segundo número entero positivo: "))
num3=int(input("Ingrese un tercer número entero positivo: "))

if num1>num2 and num2>=num3:
    print(f"{num3} es menor o igual que {num2} y {num1} es el mayor.")
elif num2>num1 and num1>=num3:
    print(f"{num3} es menor o igual que {num1} y {num2} es el mayor.")
elif num3>num2 and num2>=num1:
    print(f"{num1} es menor o igual que {num2} y {num3} es el mayor.")
else:
    print(f"{num1}, {num2} y {num3} son iguales.")
```

# ESTRUCTURAS REPETITIVAS

**FUNCIÓN RANGE**

*Sintaxis*

1. range(stop)

Genera una serie de números (0, 1, 2, 3, ..., stop-1)

Ejemplo:

range(5): 0, 1, 2, 3, 4

```python
print(list(range(5)))
```

2. range(start, stop)

[start, start+1, start+2, ..., start+stop-1]

Ejemplo:

range(1, 3) -> 1, 2

range(20, 25) -> 20, 21, 22, 23, 24

```python
print(list(range(1, 3)))
```

3. range(start, stop, step)

(start, start+step, start+step*2, start+step*3, ..., start+step*10 < stop-1)

Ejemplo:

range(2, 10, 3) -> 2, 5, 8

range(2, 10, 1) -> 2, 3, 4, 5, 6, 7, 8, 9

range(18, 1, -4) -> 18, 14, 10, 6, 2

```python
print(list(range(2, 10, 3)))
print(list(range(2, 10, 1)))
print(list(range(18, 1, -4)))
```

**CICLOS PARA O FOR**

Son una estructura iterativa que repite un bloque de instrucciones cierta cantidad de veces.

*Sintaxis*

for variable in range(rango):

    # Bloque que se quiere repetir.

Ejemplo:

Escribir un programa que grite "GOL" 10 veces.

```python
for gol in range(10):
    print("GOL", end="!! ")
```

```python
# Ejemplo 1: Imprimir los números en pares desde 0 hasta el que indique el usuario.

num=int(input("Ingrese hasta donde quiera los pares: "))
for par in range(0, num, 2):
    print(f"{par}", end=", ")
```

```python
# Ejemplo 2: Hacer un programa que imprima la tabla de multiplicar de 1 a 9, de un número indicado por el usuario.

num=int(input("Ingrese un número: "))
for i in range(1, 10):
    print(f"{num} x {i} = {i*num}")
```

```python
# Ejemplo 3: Calcular la suma del primer millon de números.

suma=0 # Suma es una variable de tipo acumulador
for i in range(1, 1000_001):
    suma=suma+i
print("{:,}".format(suma))
```

```python
# Ejemplo 4: Calcular la suma de los pares de 1 hasta 1 millón.

suma=0 # Suma es una variable de tipo acumulador
for i in range(0, 1_000_001):
    if(i%2==0):
        suma=suma+i
print("{:,}".format(suma))

suma=0 # Suma es una variable de tipo acumulador
for i in range(0, 1_000_001, 2):
    suma=suma+i
print("{:,}".format(suma))
```

```python
# Ejemplo 5: Calcular el factorial de un número.
# Se define como: factorial(n): n!=1*2*3*4...*n
# factorial(0): 0!=1

n=int(input("Ingrese un número: "))
if n>0:
    fact=1
    for i in range(2, n+1):
        fact=fact*i
    print(f"{n}!={fact}")
elif n==0:
    print("0!=1")
else:
    print("No se puede calcular el factorial porque el número es inválido.")
```

**EJERCICIO 2**

Se requiere un programa para obtener el promedio de las notas (1 a 10) de 4 estudiantes.

```python
suma=0
for i in range(4):
    nota=float(input("Ingrese una nota (1-10): "))
    suma=suma+nota
promedio=suma/4
print(f"El promedio de las notas es igual a: {promedio}")
```

**EJERCICIO 3**

Desarrolle un programa que calcule el mejor estudiante, de acuerdo con el promedio de sus 3 notas. La cantidad de estudiantes en el curso son 4.

```python
mejorEstudiante=""
mejorPromedio=0

for e in range(4):
    nombre=input("Ingrese el nombre del estudiante: ")
    nota1=float(input("Ingrese la primera nota: "))
    nota2=float(input("Ingrese la segunda nota: "))
    nota3=float(input("Ingrese la tercera nota: "))
    promedio=(nota1+nota2+nota3)/3
    if promedio>mejorPromedio:
        mejorPromedio=promedio
        mejorEstudiante=nombre

print(f"El mejor estudiante es {mejorEstudiante}.")
print(f"Promedio {mejorPromedio:.2f}")
```
