```python
import tensorflow as tf
import numpy as np
```

```python
radios = np.random.uniform(1, 20, 500)
areas = np.pi * (radios**2)
```

```python
oculta_1 = tf.keras.layers.Dense(units=64, activation='relu', input_shape=[1])
oculta_2 = tf.keras.layers.Dense(units=64, activation='relu')
salida = tf.keras.layers.Dense(units=1)

modelo = tf.keras.Sequential([oculta_1, oculta_2, salida])
```

```python
modelo.compile(
    optimizer = tf.keras.optimizers.Adam(0.1),
    loss = 'mean_squared_error'
)
```

```python
print("Entrenando...")

historial = modelo.fit(radios, areas, epochs=2000, verbose=False)

print("Entrenamiento finalizado")
```

```python
import matplotlib.pyplot as plt
plt.xlabel("# Epoca")
plt.ylabel("Magnitud de perdida")
plt.plot(historial.history["loss"])
```

```python
print("Predicción")
resultado = modelo.predict(np.array([15]))
print(f"El área del circulo es igual a: {resultado}")
```
