```python
import tensorflow as tf
import numpy as np
```

```python
inversion = np.array([1, 2, 5, 10, 20, 30, 50], dtype='float')
ventas = np.array([7, 12, 27, 52, 102, 152, 252], dtype='float')
```

```python
oculta_1 = tf.keras.layers.Dense(units=3, input_shape=[1])
oculta_2 = tf.keras.layers.Dense(units=3)
salida = tf.keras.layers.Dense(units=1)

modelo = tf.keras.Sequential([oculta_1, oculta_2, salida])
```

```python
modelo.compile(
    optimizer = tf.keras.optimizers.Adam(0.01),
    loss = 'mean_squared_error'
)
```

```python
print("Entrenando...")

historial = modelo.fit(inversion, ventas, epochs=500, verbose=False)

print("Entrenamiento finalizado")
```

```python
import matplotlib.pyplot as plt
plt.xlabel("# Epoca")
plt.ylabel("Magnitud de perdida")
plt.plot(historial.history["loss"])
```

```python
print("Predicción")
resultado = modelo.predict(np.array([85]))
print(f"La proyección es de: {resultado} ventas")
```
