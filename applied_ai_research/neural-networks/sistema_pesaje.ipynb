```python
import tensorflow as tf
import numpy as np
```

```python
kilogramos = np.array([1, 5, 10, 20, 50, 75, 100], dtype='float')
libras = np.array([2.20, 11.02, 22.05, 44.09, 110.23, 165.35, 220.46], dtype='float')
```

```python
capa_oculta_1 = tf.keras.layers.Dense(units=3, input_shape=[1])
capa_oculta_2 = tf.keras.layers.Dense(units=3)
salida = tf.keras.layers.Dense(units=1)

modelo = tf.keras.Sequential([capa_oculta_1, capa_oculta_2, salida])
```

```python
modelo.compile(
    optimizer=tf.keras.optimizers.Adam(0.1),
    loss='mean_squared_error'
)
```

```python
print("Entrenando...")
historial = modelo.fit(kilogramos, libras, epochs=200, verbose=False)

print("Fin del entrenamiento")
```

```python
import matplotlib.pyplot as plt
plt.xlabel("# Epoca")
plt.ylabel("Magnitud de pérdida")
plt.plot(historial.history["loss"])
```

```python
print("Predicción")
resultado = modelo.predict(np.array([150.0]))
print(f"El resultado es: {resultado}lb")
```
